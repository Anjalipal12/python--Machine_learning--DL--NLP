### Abstraction In OOPS
#### Introduction to Abstraction in Object-Oriented Programming
In this lecture, we continue our discussion on Python by exploring a new topic called abstraction, which is one of the most important core pillars in object-oriented programming (OOP).

#### What is Abstraction?
Abstraction is the concept of hiding the complex implementation details and showing only the necessary features of an object. This approach helps in reducing programming complexity and effort.

#### Real-World Examples of Abstraction
To better understand abstraction, consider the example of a washing machine. When using a washing machine, you have options to start it, set the timer, and use the dryer function. With just the click of a button, you can perform these operations.

The buttons visible on the washing machine represent the features accessible to you. However, the complex internal processes, such as how the washing cycle operates after pressing start, are hidden from the user. This is a practical example of abstraction.

Other examples include mobile phones, AC remotes, laptops, and desktops. You see buttons or options, such as shutting down the system or opening an application, but the complex processes happening in the background, like memory allocation or processor usage, are hidden from the user. This encapsulation of complexity is abstraction.

#### Implementing Abstraction in Python
Let's explore how abstraction is implemented in Python using the Abstract Base Class (ABC) module.

```python
from abc import ABC, abstractmethod
```

#### Defining an Abstract Class
An abstract class is defined by inheriting from the ABC class. For example, consider a Vehicle class that serves as an abstract base class.

```python
class Vehicle(ABC):
    def drive(self):
        print("The vehicle is used for driving.")
```

#### Defining an Abstract Method
An abstract method is defined within the abstract class using the @abstractmethod decorator. It is an empty method that must be implemented by any subclass inheriting from the abstract class.

```python
class Vehicle(ABC):
    def drive(self):
        print("The vehicle is used for driving.")

    @abstractmethod
    def start_engine(self):
        pass
```

#### Implementing the Abstract Method in a Derived Class
A child class inheriting from the abstract base class must implement the abstract method. For example, a Car class inherits from Vehicle and provides its own implementation of the start_engine method.

```python
class Car(Vehicle):
    def start_engine(self):
        print("Car engine started.")
```

#### Demonstrating Abstraction with a Function
We can create a function to operate a vehicle by calling its start_engine method. This demonstrates abstraction by exposing only the necessary feature and hiding the complex implementation.

```python
def operate_vehicle(vehicle):
    vehicle.start_engine()
```

```python
car = Car()
operate_vehicle(car)
```

#### Using Both Normal and Abstract Methods
The derived class can also use normal methods defined in the abstract base class. For example, the drive method can be accessed by the Car instance.

```python
car.drive()
```

#### Summary
In summary, abstraction is the concept of hiding complex implementation details and showing only the necessary features of an object. In Python, abstraction is implemented using abstract base classes and abstract methods. Abstract methods are defined with the @abstractmethod decorator and must be implemented by derived classes. This approach helps in reducing programming complexity and effort by exposing only essential functionalities.
