### What Are Magic Methods?
Magic methods in Python are also known as dunder methods, which means double underscore methods. These are special methods that start and end with double underscores. Magic methods enable you to define the behavior of an object for built-in operations such as arithmetic operations, comparisons, and more.

Some examples of magic methods include __init__, which initializes a new instance of a class (the constructor), and __str__, which returns a string representation of an object.

Viewing Magic Methods with dir()
You can view all the magic methods available for a class by using the dir() function. For example, if you create a class called Person and then create an object of this class, you can use dir(Person) to see all the magic methods.

```python
class Person:
    pass

person = Person()
dir(Person)
```

#### Using and Overriding Magic Methods
Magic methods are used to define the behavior of an object for built-in operations. For example, the __str__ method returns a string representation of an object. By default, printing an object will display a message from this function.

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

person = Person('Krish', 34)
print(person)
```

By default, printing the person object will display the memory location. To override this and display a custom message, you can override the __str__ method.

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def __str__(self):
        return f"{self.name}, {self.age} years old"

person = Person('Krish', 34)
print(person)
```

Now, when you print the person object, it will display the custom message defined in the __str__ method.

#### The repr Magic Method
Another important magic method is __repr__, which returns an official string representation of an object. You can override this method to display a custom string.

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def __repr__(self):
        return f"Person(name={self.name}, age={self.age})"

person = Person('Krish', 34)
print(repr(person))
```

By calling repr(person), you will see the custom string representation as defined in the __repr__ method.
