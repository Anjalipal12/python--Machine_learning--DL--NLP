### Operator Overloading In Python
#### Introduction to Operator Overloading in Python
In this lecture, we continue our discussion on Python by exploring operator overloading. Previously, we discussed magic methods, which are special methods in Python that allow customization of class behavior. Among these, the __eq__ method is a default method used for comparison operations. For example, when using the double equal sign == for comparison, overriding the __eq__ method allows us to customize this behavior.

Similarly, operators like addition (+) and subtraction (-) can be overloaded to define custom behavior. We will see how to override these operators, including greater than (>), to customize their functionality.

#### Magic Methods for Operator Overloading
Python provides several magic methods to overload operators:

__add__: for addition using the + operator.
__sub__: for subtraction using the - operator.
__mul__: for multiplication using the * operator.
__truediv__: for division using the / operator.
__eq__: for equality comparison using ==.
__lt__: for less than comparison using <.
__gt__: for greater than comparison using >.
We can override these methods in our classes to customize how these operators behave with our objects.

#### Defining a Vector Class with Operator Overloading
Let's define a class called Vector to demonstrate operator overloading. The class will have two attributes, x and y, representing coordinates.

We start by defining the __init__ method to initialize these attributes.

```python
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y
```

#### Overloading the Addition Operator
We overload the addition operator by defining the __add__ method. It takes two parameters: self and other, where other is another Vector instance. The method returns a new Vector whose x and y are the sums of the corresponding coordinates.

```python
def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)
```

#### Overloading the Subtraction Operator
Similarly, we overload the subtraction operator by defining the __sub__ method. It subtracts the x and y values of the other vector from self.

```python
def __sub__(self, other):
        return Vector(self.x - other.x, self.y - other.y)
```

#### Overloading the Equality Operator
To customize equality comparison, we override the __eq__ method. It returns True if both the x and y coordinates of the two vectors are equal, otherwise False.

```python
def __eq__(self, other):
        return self.x == other.x and self.y == other.y
```

#### Representing the Vector as a String
To display the vector in a readable format, we override the __repr__ method. It returns a string representation of the vector showing its x and y coordinates.

```python
def __repr__(self):
        return f"Vector({self.x}, {self.y})"
```

#### Using the Vector Class with Overloaded Operators
We create two vector objects v1 and v2 with coordinates (2, 3) and (4, 5) respectively. Then, we demonstrate addition, subtraction, and multiplication operations using the overloaded operators.

```python
v1 = Vector(2, 3)
v2 = Vector(4, 5)

print(v1 + v2)  # Output: Vector(6, 8)
print(v1 - v2)  # Output: Vector(-2, -2)
```

#### Overloading Multiplication with a Scalar
We can also overload multiplication to multiply a vector by a scalar value. The __mul__ method takes self and other where other is a number. It returns a new vector with each coordinate multiplied by the scalar.

```python
def __mul__(self, other):
        return Vector(self.x * other, self.y * other)
```

```python
print(v1 * 3)  # Output: Vector(6, 9)
```

```python
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    # Addition
    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)

    # Subtraction
    def __sub__(self, other):
        return Vector(self.x - other.x, self.y - other.y)

    # Equality
    def __eq__(self, other):
        return self.x == other.x and self.y == other.y

    # Multiplication with scalar
    def __mul__(self, scalar):
        return Vector(self.x * scalar, self.y * scalar)

    # String representation
    def __repr__(self):
        return f"Vector({self.x}, {self.y})"

# Objects
v1 = Vector(2, 3)
v2 = Vector(4, 5)

# Operations
print(v1 + v2)  # Vector(6, 8)
print(v1 - v2)  # Vector(-2, -2)
print(v1 * 3)   # Vector(6, 9)
print(v1 == v2) # False
```

#### Summary
By overloading operators such as addition, subtraction, multiplication, and equality, we can define intuitive operations for custom classes like vectors. This allows us to use standard operators with objects in a natural and readable way.
