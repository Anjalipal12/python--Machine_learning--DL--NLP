### Importing NumPy
To use NumPy, you need to import it. The common convention is to import it as np, which serves as an alias and allows you to use np. instead of typing numpy. everywhere.

```python
import numpy as np
```

#### Creating Arrays
Let us create a one-dimensional array (1D array) using NumPy. We use the np.array function, which can take a list or tuple as input.

```python
arr1 = np.array([1, 2, 3, 4, 5])
print(arr1)
```

#### Array Type and Shape
Although the array may look like a list, it is actually a NumPy ndarray. You can check its type and shape using the following commands:

```python
print(type(arr1))
print(arr1.shape)
```

#### Reshaping Arrays
If you want to convert a 1D array into a 2D array, you can use the reshape method. For example, reshaping to (1, 5) means one row and five columns. The total number of elements must remain the same.

```python
arr2 = arr1.reshape(1, 5)
print(arr2)
print(arr2.shape)
```

#### Creating 2D Arrays Directly
You can also create a two-dimensional array directly by passing a nested list to np.array.

```python
arr2 = np.array([[1, 2, 3, 4, 5], [2, 3, 4, 5, 6]])
print(arr2)
print(arr2.shape)
```

#### Using Inbuilt Functions to Create Arrays
NumPy provides several inbuilt functions to create arrays quickly, such as arange, ones, and eye.

```python
arr = np.arange(0, 10, 2)
print(arr)
arr_reshaped = arr.reshape(5, 1)
print(arr_reshaped)
```
