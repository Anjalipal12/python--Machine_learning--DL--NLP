### Data Visualization With Seaborn
#### Introduction to Seaborn
In this video, we will discuss a new library called Seaborn, which is used for data visualization in Python. Seaborn is a Python visualization library based on matplotlib. It provides a higher-level interface for drawing attractive and informative statistical graphics. Seaborn helps in creating complex visualizations with just a few lines of code. In this lesson, we will cover the basics to advanced features of Seaborn, including creating various types of plots and customizing them.

### Importing Seaborn
To use Seaborn, import it and assign an alias. The common alias is sns.

```python
import seaborn as sns
```

#### Loading Built-in Datasets
Seaborn provides several inbuilt datasets. You can load a dataset using the load_dataset function. For example, to load the 'tips' dataset:

```python
tips = sns.load_dataset('tips')
print(tips)
```

The 'tips' dataset contains features such as total_bill, tip, sex, smoker, day, time, and size. This dataset includes both categorical and numerical data, making it suitable for various types of plots.

#### Basic Plotting with Seaborn
#### Scatter Plot
To create a scatter plot of total_bill versus tip, you can use the following code:

```python
import matplotlib.pyplot as plt
sns.scatterplot(x='total_bill', y='tip' , data=tips)
plt.title('Scatter Plot of Total Bill vs Tip')
plt.show()
```

This scatter plot shows the distribution and relationship between total_bill and tip.

#### Line Plot
To create a line plot of total_bill by size, use:

```python
sns.lineplot(x='size' , y='total_bill', data=tips)
plt.title('Line Plot of Total Bill by Size')
plt.show()
```

The line plot displays the distribution of total bills based on the size of the group dining at the restaurant.
#### Categorical Plots
Seaborn allows for visualization of categorical variables such as sex, smoker, day, and time.

#### Bar Plot
To create a bar plot of total bill by day:

```python
sns.barplot(x='day', y='total_bill' , data=tips)
plt.title('Plot of Total Bill by Day')
plt.show()
```

#### Box Plot
Box plots are useful for identifying outliers and understanding the distribution of data. To create a box plot of total bill by day:

```python
sns.boxplot(x='day', y='total_bill', data=tips)
plt.title('Box Plot of Total Bill by Day')
plt.show()
```

The box plot displays the minimum, 25th percentile, median (50th percentile), 75th percentile, and maximum values, as well as outliers for each day.

#### Violin Plot
Violin plots show the distribution of the data. To create a violin plot:

```python
sns.violinplot(x='day', y='total_bill', data=tips)
plt.title('Violin Plot of Total Bill by Day')
plt.show()
```

The violin plot provides a visual representation of the distribution and density of the total bill for each day.

#### Histograms and KDE Plots
Seaborn can also be used to create histograms and kernel density estimation (KDE) plots.

#### Histogram
To create a histogram of total_bill with 10 bins and show the KDE curve:

```python
sns.histplot(tips['total_bill'], bins=10, kde=True)
plt.title('Histogram of total bill with KDE')
plt.show()
```

Setting kde=True adds a smooth curve representing the distribution of the data.

#### KDE Plot
To focus solely on the KDE plot:

```python
sns.kdeplot(tips['total_bill'],fill=True)
plt.title('KDE Plot of Total Bill')
plt.show()
```

The fill=True parameter fills the area under the KDE curve, providing a clearer visualization of the distribution.
#### Pair Plot
Pair plots are useful for visualizing relationships between all numerical variables in the dataset. To create a pair plot:

```python
sns.pairplot(tips)
plt.show()
```

The pair plot shows the relationships between each pair of numerical features, with histograms on the diagonal. Categorical variables are not included in the pair plot.

#### Correlation and Heatmap
To visualize the correlation between numerical features, use a heatmap. First, select only the numerical columns and compute the correlation matrix:

```python
numerical_features = ['total_bill','tip','size']
correlation = tips[numerical_features].corr()
print(correlation)
```

The correlation matrix shows the relationship between each pair of numerical features. For example, if total_bill increases by one unit, tip increases by approximately 0.67 units.

To visualize this correlation matrix as a heatmap:

```python
sns.heatmap(correlation, annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```

The heatmap uses color to represent the strength of the correlation between variables. Red indicates a strong positive correlation, while blue indicates a weaker or negative correlation.

#### Customizing Plots with Matplotlib and Seaborn
You can customize plots further by combining Seaborn with matplotlib. For example, to plot total sales by product using a custom dataset:

```python
import pandas as pd
Sales_target_df = pd.read_csv('Sales_target.csv')
plt.figure(figsize=(10,6))
sns.barplot(x='Category', y='Target', data=Sales_target_df, estimator=sum)
plt.title('Total Slaes by Product')
plt.xlabel('Category')
plt.ylabel('Target')
plt.show()
```

This bar plot shows the total revenue for each product category. The estimator parameter allows you to specify aggregation functions such as sum or mean.

Similarly, you can plot total sales by region:

```python
sns.barplot(x='Category', y='Target', data=Sales_target_df, estimator=sum)
plt.title('Total Sales by Region')
plt.xlabel('Category')
plt.ylabel('Target')
plt.show()
```

This bar plot displays the total sales for each region, making it easy to compare sales performance across different regions.
