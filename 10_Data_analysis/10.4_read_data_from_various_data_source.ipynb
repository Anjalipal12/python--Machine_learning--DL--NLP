### Reading Data From Various Data Source Using Pandas
In this tutorial, we will continue our discussion on pandas. We will explore how to read data from different sources using some amazing built-in functions available in pandas.

```python
import pandas as pd
from io import StringIO

json_data = '''
[{"employee_name": "James" , "email": "james@gmail.com", "job_profile": "Team leader"}, {"employee_name": "Michael", "email": "michael@gmail.com" , "job_profile": "Senior Developer"}]'''
df = pd.read_json(StringIO(json_data))
```

#### Reading JSON Data
Here, we have some JSON data representing employee details such as name, email, and job profile. We convert this JSON string into a pandas DataFrame using pd.read_json() combined with StringIO to simulate a file-like object. The resulting DataFrame displays the employee information in tabular form.

#### Converting DataFrame back to JSON
We can convert the DataFrame back into JSON format using the to_json() method. By default, the JSON output uses the orient='index' parameter, which structures the JSON with column names as keys and index-based nested objects.

```python
json_default = df.to_json()
print(json_default)

json_records = df.to_json(orient='records')
print(json_records)
```

```python
json_default = df.to_json()   #Ye DataFrame ko JSON format me convert karta hai (default column-wise format me).
                            #Means — har column ek key ban jaata hai.
print(json_default)

json_records = df.to_json(orient='records')   #Ye DataFrame ko list of dictionaries (row-wise JSON) me convert karta hai —
                                               #yaani har row ek record (dictionary) ban jaati hai.
print(json_records)
```

The orient parameter controls the JSON format:

index (default): JSON keys are column names with index-based nested objects.
records: JSON is a list of records, each record being a dictionary of column-value pairs.
Using orient='records' produces a JSON array where each element corresponds to a row in the DataFrame.

### Reading CSV Data from a URL
We can read CSV data directly from a URL using pd.read_csv(). For example, reading a wine dataset in CSV format from a web URL. If the CSV file does not contain a header row, specify header=None to assign default integer column names.

```python
url = 'https://archive.ics.uci.edu/ml/machine-learning-databases/wine/wine.data'
df_csv = pd.read_csv(url, header=None)
print(df_csv.head())
```

### Saving DataFrame to CSV
We can save the DataFrame back to a CSV file using the to_csv() method. This writes the DataFrame to a file in the current working directory.

```python
df_csv.to_csv('wine.csv', index=False)
```

### Reading HTML Tables
Pandas can read tables from HTML pages using pd.read_html(). This function returns a list of DataFrames, each corresponding to a table found in the HTML content. However, it requires additional dependencies such as lxml, html5lib, and beautifulsoup4 to parse HTML.

```python
# Install dependencies if not already installed
!pip install lxml html5lib beautifulsoup4
```

```python
url_html = 'https://www.fdic.gov/resources/resolutions/bank-failures/failed-bank-list/'
df_list = pd.read_html(url_html)
print(len(df_list))  # Number of tables found
print(df_list[0].head())  # Display first table
```

#### Reading Specific Tables with Match Keyword
pd.read_html() supports a match parameter to filter tables containing a specific keyword. This helps extract the desired table from a page with multiple tables.

```python
!pip install requests
```

```python
import pandas as pd
import requests
from io import StringIO

# Wikipedia page URL
url = 'https://en.wikipedia.org/wiki/Mobile_country_code'

# Fake browser headers bhej rahe hain
headers = {'User-Agent': 'Mozilla/5.0'}

# Page ko download karna
response = requests.get(url, headers=headers)

# HTML content ko Pandas se read karna
df_list = pd.read_html(StringIO(response.text), match='Country', header=0)

# First matching table ka first 5 rows
print(df_list[0].head())
```

### Pickle Files
Pickle files serialize Python objects into byte streams for storage or transmission. Pandas supports reading and writing DataFrames to pickle files using to_pickle() and read_pickle() methods. This is useful for saving models or data states in machine learning projects.

Pickle → Python objects ko save & load karne ka fast way
Pandas → to_pickle() + read_pickle()

```python
import pandas as pd

# Assume df already exists (DataFrame)

# Save DataFrame to pickle
df.to_pickle('data.pkl')  # correct method & filename

# Load DataFrame from pickle
df_pickle = pd.read_pickle('data.pkl')  # correct variable

# Show first 5 rows
print(df_pickle.head())
```
