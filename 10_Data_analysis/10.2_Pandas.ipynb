#### Pandas - DataFrame and Series
#### Introduction to Pandas: DataFrame and Series
In this lecture, we continue our Python series, focusing on the pandas library. Pandas is a powerful data manipulation library in Python, widely used for data analysis and cleaning. It provides two primary data structures: Series and DataFrame.

#### Pandas Series
A pandas Series is a one-dimensional array-like object that can hold any data type. It is similar to a column in a table. The basic difference between a Series and a DataFrame is that a Series is one-dimensional, while a DataFrame is two-dimensional, mutable, and can contain heterogeneous data.

```python
import pandas as pd
```

#### Creating a Series from a List
You can create a Series from a list using the pd.Series function. For example, using the data [1, 2, 3, 4, 5].

```python
data = [1,2,3,4,5]
series = pd.Series(data)
print(series)
```

#### Creating a Series from a Dictionary
You can also create a Series from a dictionary. The keys become the index.

```python
data_dict = {'a': 1, 'b': 2, 'c': 3}
series_dict = pd.Series(data_dict)
print(series_dict)
```

#### Custom Index in Series
You can assign a custom index to a Series by passing the index parameter.

```python
data = [10,20,30]
index = ['a','b','b']
series_custom = pd.Series(data, index = index)
print(series_custom)
```

#### Pandas DataFrame
A DataFrame is a two-dimensional, mutable, and potentially heterogeneous tabular data structure with labeled axes (rows and columns). It is similar to a table in a database or an Excel sheet.

#### Creating a DataFrame from a Dictionary of Lists
You can create a DataFrame by passing a dictionary where the values are lists.

```python
data = {
    'name': ['Krish' , 'John' , 'Jack'],
    'age' : [25,30,45],
    'city' : ['Bangalore' , 'New York' , 'Florida']
}
df = pd.DataFrame(data)
print(df)
```

#### Creating a DataFrame from a List of Dictionaries
You can also create a DataFrame from a list of dictionaries, where each dictionary represents a row

```python
data = [
    {'name': 'Krish' , 'age': 32 , 'city':'Banglore'},
    {'name': 'John' , 'age' : 34 , 'city':'Banglore'},
    {'name': 'jack' , 'age': 32 , 'city': 'Banglore'}
]
df = pd.DataFrame(data)
print(df)
```

#### Reading Data from CSV Files
Pandas can read data from CSV files using the read_csv function. This is commonly used for larger datasets.

```python
df = pd.read_csv('car_prices.csv')
print(df.head())
```

#### Features of DataFrame

2D data structure — rows and columns.

Heterogeneous data — har column alag type ka ho sakta hai (int, float, string).

Labels — rows and columns ke liye labels hota hai.

Flexible indexing — .loc, .iloc, .at, .iat use karke specific rows/columns access kar sakte hain.

#### Accessing Data in DataFrames
You can access columns directly by their names, or use .loc, .iloc, .at, and .iat for more specific access.

```python
df['name']  # Returns all names
```

```python
df.loc[0]  
#.loc uses labels (names of rows or columns).
```

```python
df.iloc[0, 0]  
# .iloc uses integer positions (like 0,1,2…).
```

```python
df.at[1, 'age']
```

```python
df.iat[0, 1]
# Access single value quickly with .iat
```

```python
df.iat[2, 2]
```

#### Data Manipulation with DataFrames
You can add new columns, remove columns, and update values in DataFrames.

```python
df['salary'] = [50000, 60000, 70000]
print(df)
```

```python
df.drop('salary', axis=1 , inplace = True)
print(df)
```

```python
df['age'] = df['age'] + 1
print(df)
```

```python
df.drop(0, inplace=True)
print(df)
```

#### DataFrame Attributes and Methods
You can use various DataFrame methods to explore and summarize your data, such as .head(), .describe(), and checking data types.

```python
print(df.head())
```

```python
print(df.dtypes)
```

```python
print(df.describe())
```
