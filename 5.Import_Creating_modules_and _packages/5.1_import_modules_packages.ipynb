### Import Modules And Packages In Python
#### Introduction to Python Modules and Packages
Python is an open-source programming language that includes many packages. These packages contain numerous built-in functions, allowing you to use them without writing code from scratch. Modules and packages help organize and reuse code efficiently.

```python
import math
```

The math package is a module that contains many built-in mathematical functions. To import any package, you simply write import followed by the package name.

```python
math.sqrt(16)
```

For example, the sqrt function in the math package calculates the square root of a number. Executing math.sqrt(16) returns 4.

```python
from math import sqrt, pi
```

You can also import specific functions from a package using the syntax from package import function. For instance, importing sqrt and pi from math allows you to use them directly without the math. prefix.

```python
sqrt(16)
sqrt(25)
print(pi)
```

After importing, you can call these functions directly. For example, sqrt(16) returns 4, sqrt(25) returns 5, and pi is approximately 3.141.

```python
import numpy as np
```

NumPy is a powerful package used for array operations. It is common to import it with an alias, such as np, to simplify code by avoiding repeated long package names.

```python
import numpy as np
arr = np.array([1, 2, 3, 4])
print(arr)
```

```python
from math import *
```

Using from math import * imports all functions from the math package, allowing you to use any function directly without prefixing with math..

```python
print(sqrt(16))
print(pi)
```

After importing all functions, you can call sqrt(16) and pi directly, which will output 4 and approximately 3.141 respectively.

#### Creating Custom Packages and Modules
You can create your own custom packages by organizing your Python files in folders with a special __init__.py file. This file can be empty but is required to mark the folder as a package.

```python
# This is a special file
# __init__.py

"""
This file initializes the package namespace.
"""
```

Inside your package folder, create Python files (modules) with functions. For example, a file maths.py with an addition function:

```python
def addition(a, b):
    return a + b
```

With the __init__.py file present, the folder becomes a package. You can import your custom functions using the syntax from package.module import function.

```python
from package.maths import addition

result = addition(2, 3)
print(result)
```

Alternatively, you can import the module and call the function with the module prefix:

```python
from package import maths

result = maths.addition(2, 3)
print(result)
```

You can add multiple functions to your module, for example, a subtraction function:

```python
def subtraction(a, b):
    return a - b
```
