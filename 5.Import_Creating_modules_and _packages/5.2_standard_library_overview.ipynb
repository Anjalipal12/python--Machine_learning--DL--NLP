### Standard Library Overview
#### Introduction to Python Standard Library
Python provides a vast collection of libraries and packages that are open source and can be used in your projects or development work. In this lecture, we will discuss an overview of some important standard libraries in Python.

#### Working with Arrays
The array library is an important module that helps in creating arrays. You can define an array variable using the array.array function, which requires two parameters: a type code (such as 'i' for integers) and an initializer (a list of values).

```python
import array
arr = array.array('i', [1, 2, 3, 4])
print(arr)
```

#### Mathematical Operations
The math module provides various mathematical functions. For example, you can use math.sqrt() to find the square root of a number and math.pi to get the value of pi.

```python
import math
print(math.sqrt(16))
print(math.pi)
```

#### Generating Random Numbers
The random module is commonly used for generating random numbers. Two important functions are randint, which returns a random integer within a specified range, and choice, which selects a random element from a collection.

```python
import random
print(random.randint(1, 10))
fruits = ['apple', 'banana', 'cherry']
print(random.choice(fruits))
```

#### File and Directory Operations with OS Module
The os module is used for file and directory access. You can get the current working directory and create new directories using this module.

```python
import os
print(os.getcwd())
os.mkdir('test_dir')
```

High-Level File Operations with shutil
For high-level operations on files and collections of files, the shutil module is used. For example, you can copy files from one location to another.

```python
import shutil
shutil.copyfile('source.txt', 'destination.txt')
```

#### Data Serialization with JSON
Data serialization, such as converting a dictionary into JSON, is important when working with APIs or databases. The json module provides dumps to convert a dictionary to a JSON string and loads to convert a JSON string back to a dictionary.

```python
import json
data = {'name': 'Krish', 'age': 25}
json_str = json.dumps(data)
print(json_str)
print(type(json_str))
parsed_data = json.loads(json_str)
print(parsed_data)
print(type(parsed_data))
```

### Working with CSV Files
The csv module is used for reading and writing CSV files. You can write rows to a CSV file and then read them back using the appropriate reader and writer functions.

```python
import csv
with open('example.csv', 'w', newline='') as file:
    writer = csv.writer(file)
    writer.writerow(['name', 'age'])
    writer.writerow(['Krish', 32])
with open('example.csv', 'r') as file:
    reader = csv.reader(file)
    for row in reader:
        print(row)
```

#### Date and Time Operations
The datetime module allows you to work with dates and times. You can get the current date and time, and calculate time differences using timedelta.

```python
from datetime import datetime, timedelta
now = datetime.now()
print(now)
yesterday = now - timedelta(days=1)
print(yesterday)
```

#### Program Timing with Time Module
The time module can be used to measure time and make the program sleep for a specified number of seconds.

```python
import time
print(time.time())
time.sleep(2)
print(time.time())
```

### Regular Expressions with re Module
The re module is used for regular expressions, which help in matching specific patterns in text. For example, you can search for digits in a string using a pattern.

```python
import re
pattern = r'\d+'
text = 'There are 123 apples.'
match = re.search(pattern, text)
if match:
    print(match.group())
```
