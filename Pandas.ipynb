Pandas Series – Interview Q&A Cheat Sheet
1️⃣ Basics

Q1: Pandas Series kya hai?
A: Series = 1D labeled array in Pandas. Data + Index (labels).

Q2: Series aur Numpy array mein difference kya hai?
A:

Numpy array: only numeric, index integer-based

Series: numeric/non-numeric data + custom labels (index) possible

Q3: Default index kya hai?
A: Jab custom index na diya ho, Pandas automatically 0,1,2,… integer index assign karta hai.

Q4: Custom index ka use kyun?
A: Data ko readable aur label-based access ke liye, jaise 'A','B','C' instead of 0,1,2

2️⃣ Indexing / Access

Q5: s[0] aur s['A'] mein kya difference?
A:

s[0] → integer position (default index)

s['A'] → label-based access (custom index 'A')

Q6: s[-1] hamesha last element return karega?
A:

Default index → last element return karega

Custom index → ERROR, kyunki -1 label exist nahi karta

Q7: .loc vs .iloc?
A:

.loc[label] → label-based access

.iloc[position] → position-based access

Q8: Negative indexing kaise kaam karta hai?
A:

Default integer index → last element = s[-1]

Custom index → use .iloc[-1]

3️⃣ Slicing / Modification

Q9: Series me slice assignment kaise karte hain?
A:

s[:3] = 20   # first 3 elements ka value = 20


Q10: Single element modify kaise karte hain?

s['A'] = 100  # label 'A' ka value change

4️⃣ Vectorized Operations

Q11: Arithmetic operations kaise kaam karte hain?

s + 10    # add 10 to each element
s / 10    # divide each element by 10
s * 10    # multiply each element by 10


Q12: Boolean filtering example?

s[s < 30]  # returns elements where value < 30

5️⃣ Aggregation / Statistics

Q13: Sum, mean, max, min kaise nikalte hain?

s.sum()
s.mean()
s.max()
s.min()


Q14: NaN values ke saath aggregation ka behavior?
A: Pandas automatically ignore NaN in aggregation functions.

6️⃣ Series Arithmetic & Alignment

Q15: s1 + s2 kaise kaam karta hai?
A:

Index same → corresponding elements add honge

Index mismatch → result = NaN for mismatched labels

Example:

s1 = pd.Series([20,20,20], index=['A','B','C'])
s2 = pd.Series([34,45,78], index=['A','B','C'])
s1 + s2
# A 54, B 65, C 98

7️⃣ Errors / Debugging

Q16: pd.series([...]) me error kyun?
A: Series ka S capital hona chahiye → pd.Series([...])

Q17: s1[-2] custom index me kyun error deta hai?
A: Pandas direct indexing by default label-based hota hai.

Custom labels me -2 exist nahi karta → ERROR

Correct: s1.iloc[-2]

✅ Extra Tip for Notes:

Always remember: Default index = 0,1,2…

Custom index → use .loc

Integer position → use .iloc

Boolean filtering, vectorized operations, aggregation → common in interviews

| Access                 | Result                       |

| `df[['salary','age']]` | DataFrame                    |
| `df.iloc[:,0]`         | First column as Series       |

| `.query()`             | Condition-based filtering with variables                      |


💡 Notes Tips:

.iloc → always integers

.loc → always labels

.query() → nice for readable filtering

@variable → Python variable inside query string

and / or → combine multiple conditions

Use case → dynamic filtering, user input based search

| Default `axis`        | 0 → row, 1 → column                    |                            |

| Concept / Method                     | Code Example                                                                | What it Does / Use Case                      | Kab Use Karte Hain                                                  |
| ------------------------------------ | --------------------------------------------------------------------------- | -------------------------------------------- | ------------------------------------------------------------------- |
| **Filter rows by column value**      | `df[df['country']=='india']`                                                | Select rows where condition is True          | Jab specific subset chahiye kisi column ke basis pe                 |
| **Filter multiple columns & mean**   | `df[df['country']=='india'][['salary','age']].mean()`                       | Column-wise mean for filtered rows           | Jab hume sirf numeric columns ka average chahiye                    |
| **GroupBy single column**            | `df.groupby('country')`                                                     | Group rows by column, returns GroupBy object | Jab data ko categories (e.g., country) ke basis pe analyse karna ho |
| **Get group indices**                | `df.groupby('country').groups`                                              | Dict of `{group_name: row_indices}`          | Kab row indices dekhni ho per group                                 |
| **Aggregate after group**            | `df.groupby('country')[['age','salary']].mean()`                            | Mean of numeric columns per group            | Jab group-wise summary chahiye                                      |
| **Group by multiple columns**        | `df.groupby(['country','gender'])[['age','salary']].mean()`                 | Mean per country-gender combination          | Jab hierarchical grouping karni ho                                  |
| **Custom aggregation**               | `df.groupby(['country','gender']).agg({'salary':'mean','height':'median'})` | Multiple aggregation functions per column    | Jab different columns pe different summary statistics chahiye       |
| **Multiple functions on one column** | `df.groupby(['country','gender']).agg({'height':['median','mean']})`        | Median & mean both for same column           | Jab ek column ka detailed analysis chahiye                          |
| **Count of rows per group**          | `df.groupby(['country','gender']).agg({'gender':'count'})`                  | Number of rows in each group                 | Jab group size check karna ho                                       |
| **Query with variables**             | `df.query("country==@country and salary>@salary")`                          | Filter rows using Python variables           | Jab dynamic / user input filtering karni ho                         |
| **Iterate over groups**              | `for country, idx in df.groupby('Country name').groups.items(): ...`        | Loop through each group                      | Jab per-group operations / plotting karni ho                        |
| **Plot per group**                   | `plt.plot(d['year'], d['Log GDP per capita'])`                              | Line plot for each country                   | Jab visual trend chahiye per group                                  |
| **Drop missing rows/columns**        | `df.dropna(axis=0, how='any')` / `df.dropna(axis=1, how='all')`             | Remove rows or columns with missing values   | Jab missing data ko ignore / remove karna ho                        |
| **Fill missing values**              | `df['salary'].fillna(df['salary'].median(), inplace=True)`                  | Replace NaN with central tendency            | Jab missing numeric data ko replace karna ho                        |
| **Check missing values**             | `df.isna().sum()` / `df.isna().sum(axis=1)`                                 | Count NaNs per column / row                  | Data cleaning step me check karne ke liye                           |
| **Access row by label**              | `df.loc['A']`                                                               | Get row by index label                       | Jab custom index row chahiye                                        |
| **Access row by position**           | `df.iloc[0]`                                                                | Get row by integer position                  | Jab position ke basis pe access chahiye                             |
| **Access column**                    | `df['salary']`                                                              | Get single column                            | Jab column-wise operation karni ho                                  |
| **Access multiple columns**          | `df[['salary','age']]`                                                      | DataFrame with selected columns              | Jab subset of columns chahiye                                       |
| **Drop column / row**                | `df.drop('new', axis=1, inplace=True)` / `df.drop(['A','B'], axis=0)`       | Delete columns or rows                       | Jab unnecessary data remove karna ho                                |

| Parameter   | Description                                   |
| ----------- | --------------------------------------------- |
| `values`    | Numeric columns to aggregate                  |
| `index`     | Rows in the output                            |
| `columns`   | Columns in the output                         |
| `aggfunc`   | Aggregation function (default = mean)         |
| Multi-index | `index=['col1','col2']` for hierarchical rows |


Difference from GroupBy:

GroupBy → summarize data, output not reshaped

Pivot Table → summarize + reshape rows x columns
