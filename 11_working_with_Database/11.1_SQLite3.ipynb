### Crud Operation With SQLite3 And Python
#### Introduction to SQL and SQLite
In this video, we are going to discuss SQL and SQLite. First, we will understand the difference between SQL and SQLite. SQL stands for Structured Query Language, which is a standard language for managing and manipulating relational databases. SQLite, on the other hand, is a self-contained, serverless, and zero-configuration database engine that is widely used for embedded data systems.

#### Differences Between SQL and SQLite
SQL is a complete application that helps manage relational databases and requires significant configuration, including setting up a server. SQLite is a lightweight version of SQL that does not require any configuration. You can create a database and store it locally, making it widely used for embedded systems.

#### Importing SQLite3 in Python
By default, in Python 3, SQLite3 comes installed. To import it, use the following command:

```python
import sqlite3
```

#### Creating and Connecting to a Database
To create and connect to an SQLite database, use the connect function. For example, to create a database named example.db:

```python
connection = sqlite3.connect('example.db')
```

After executing this, a file named example.db will be created in your folder location. This demonstrates that SQLite databases can be created and stored locally without any configuration.

#### Creating a Cursor Object
A cursor object is required to iterate through the tables and rows. To create a cursor:

```python
cursor = connection.cursor()
```

#### Creating a Table in SQLite
To create a table named employees with columns for id, name, age, and department, use the following SQL command with the cursor object:

```python
cursor.execute('''
               CREATE TABLE IF NOT EXISTS employees (
               id INTEGER PRIMARY KEY,
               name TEXT NOT NULL,
               age INTEGER,
               department TEXT)''')
```

After creating the table, commit the changes to save them:

```python
connection.commit()
```

#### Inserting Data into the Table
To insert data into the employees table, use the following commands:

```python
cursor.execute("INSERT INTO employees (name,age,department) VALUES ('Krish', 32 , 'Data scientist')")
cursor.execute("INSERT INTO employees (name,age,department) VALUES ('Bob' , 25 , 'Engineering')")
cursor.execute("INSERT INTO employees (name,age,department) VALUES ('Charlie',35,'Finance')")
connection.commit()
```

#### Querying Data from the Table
To query all data from the employees table and print the results:

```python
cursor.execute('SELECT * FROM employees')
rows = cursor.fetchall()
for row in rows:
    print(row)
```

#### Updating Data in the Table
To update the age of an employee named Krish to 34:

```python
cursor.execute("UPDATE employees SET age = 34 WHERE name = 'Krish' ")
connection.commit()
```

#### Deleting Data from the Table
To delete the record of the employee named Bob:

```python
cursor.execute("DELETE FROM employees WHERE name = 'Bob'")
connection.commit()
```

#### Bulk Insertion of Sales Data
To insert multiple records at once, use the executemany method. Prepare your data as a list of tuples:

```python
Sales_target = [
    ('2023-01-01', 'ProductA', 100, 'North'),
    ('2023-01-02', 'ProductB', 150, 'South'),
    ('2023-01-03', 'ProductC', 200, 'East'),
    ('2023-01-04', 'ProductD', 250, 'West')
]

cursor.executemany(
    'INSERT INTO sales (Month_of_order_date, Category, Target, region) VALUES (?, ?, ?, ?)',
    Sales_target
)
connection.commit()
```

#### Querying Sales Data
To retrieve and print all records from the sales table:

```python
cursor.execute('SELECT * FROM sales')
rows = cursor.fetchall()
for row in rows:
    print(row)
```

### Closing the Database Connection
To permanently close the connection to the database:

```python
connection.close()
```

If you try to execute any cursor commands after closing the connection, you will get an error stating that you cannot operate on a closed database.
