#### Managing a To-Do List with Python Lists
The first example demonstrates managing a to-do list. We will create a list to keep track of all the tasks planned for the day. For instance, after opening an app, you might select tasks such as buying groceries, cleaning the house, and paying bills. As new tasks arise, they can be added to the list. Once a task is completed, it can be removed from the list. The following code illustrates these operations.

```python
to_do_list = ["buy groceries", "clean the house", "pay bills"]
to_do_list.append("schedule meeting")
to_do_list.append("go for a run")
to_do_list.remove("clean the house")
if "pay bills" in to_do_list:
    print("Hey, do not forget to pay the utility bills.")
print("To do list remaining:")
for task in to_do_list:
    print(f"- {task}")
```

After executing this code, you will receive a reminder to pay the utility bills if that task is still present. The remaining tasks will be displayed, and you can see that completed tasks have been removed. In larger applications, lists are commonly used for such dynamic task management due to their flexibility in adding and removing items.

### Organizing Student Grades Using Lists
Another example involves organizing student grades. Suppose you have a list of grades and want to add new grades, calculate the average, and find the highest and lowest grades. The following code demonstrates these operations.

```python
grades = [85, 92, 78, 90, 88]
grades.append(95)
average_grade = sum(grades) / len(grades)
highest_grade = max(grades)
lowest_grade = min(grades)
print(f"Average grade: {average_grade}")
print(f"Highest grade: {highest_grade}")
print(f"Lowest grade: {lowest_grade}")
```

By using built-in functions like sum, max, and min, you can efficiently analyze and manage student grades stored in a list. This approach is simple yet powerful for handling collections of numerical data.

#### Managing Inventory with Lists
Managing an inventory is another practical use case for lists. In an e-commerce context, inventory management is crucial. The following code shows how to add new items, remove sold items, check for stock, and display the current inventory.

```python
inventory = ["apples", "bananas", "oranges", "grapes"]
inventory.append("strawberries")
inventory.remove("bananas")
item = "oranges"
if item in inventory:
    print(f"{item.capitalize()} are in stock.")
else:
    print(f"{item.capitalize()} are out of stock.")
print("Current inventory:")
for product in inventory:
    print(f"- {product}")
```

This code allows you to manage inventory items dynamically. You can add new items as they arrive, remove items once they are sold, and check if a specific item is in stock. Lists provide a straightforward way to handle such operations, even as the scale of data increases.

### Collecting and Analyzing User Feedback
Lists can also be used to collect and analyze user feedback. The following example demonstrates how to store feedback comments and count the number of positive feedback entries containing keywords such as "great" or "excellent."

```python
feedback = [
    "Great service.",
    "Very satisfied.",
    "Could be better.",
    "Excellent experience."
]
feedback.append("Not happy with the service.")
positive_feedback_count = sum(
    1 for comment in feedback if ("great" in comment.lower() or "excellent" in comment.lower())
)
print(f"Positive feedback count: {positive_feedback_count}")
print("User feedback:")
for comment in feedback:
    print(f"- {comment}")
```

This code counts the number of positive feedback comments by checking for the presence of the words "great" or "excellent" in each comment. The feedback list is then printed, showing all collected comments. Lists are ideal for storing and processing such collections of textual data.
