### Sets in Python
Introduction to Sets in Python
In this video, we are going to discuss a new data structure in Python called sets. We will see the definition of sets, explore examples and methods, and understand the importance of sets. Sets are one of the most common data structures used after lists, dictionaries, and tuples.

### Definition and Properties of Sets
Sets are built-in data types in Python used to store collections of unique items. They are unordered, meaning the elements do not follow a specific order, and they do not allow duplicate elements. Sets are useful for membership tests, eliminating duplicate entries, and performing mathematical set operations like union, intersection, difference, and symmetric difference.

### Creating Sets
To create a set, use curly braces similar to dictionaries, but without key-value pairs. For example, to define a set with elements 1, 2, 3, 4, 5:

```python
my_set = {1, 2, 3, 4, 5}
print(my_set)
print(type(my_set))
```

To create an empty set, use the set class:

```python
empty_set = set()
print(empty_set)
print(type(empty_set))
```

You can also create a set from a list or tuple by typecasting:

```python
my_list = [1, 2, 3, 4, 5, 6]
my_set = set(my_list)
print(my_set)
```

### Uniqueness and Unordered Nature
Sets store only unique items. If you add duplicate elements, they are ignored. For example:

```python
my_set = {1, 2, 3, 4, 5, 6, 5, 6}
print(my_set)
```

#### Basic Set Operations: Adding and Removing Elements
To add elements to a set, use the add method:

```python
my_set.add(7)
print(my_set)
```

Adding the same element again does not change the set, as duplicates are ignored.

To remove an element, use the remove method. If the element is not present, it raises a KeyError:

```python
my_set.remove(3)
print(my_set)
# my_set.remove(10)  # Raises KeyError if 10 is not present
```

To avoid errors when removing an element that may not exist, use the discard method:

```python
my_set.discard(10)
print(my_set)
```

To remove a random element, use the pop method. It removes and returns an arbitrary element:

```python
removed_element = my_set.pop()
print(removed_element)
print(my_set)
```

To clear all elements from a set, use the clear method:

```python
my_set.clear()
print(my_set)
```

#### Membership Test
To check if an element is present in a set, use the in operator:

```python
my_set = {1, 2, 3, 4, 5}
print(3 in my_set)
print(10 in my_set)
```

#### Mathematical Set Operations
Union
Union combines all elements from two sets.

```python
set1 = {1, 2, 3, 4, 5, 6}
set2 = {4, 5, 6, 7, 8, 9}
union_set = set1.union(set2)
print(union_set)
```

#### Intersection
Intersection returns only the common elements from both sets.

#### Intersection
Intersection returns only the common elements from both sets.

```python
intersection_set = set1.intersection(set2)
print(intersection_set)
```

You can also update a set with the intersection using intersection_update:

```python
set1.intersection_update(set2)
print(set1)
```

#### Difference 
Difference returns elements present in the first set but not in the second.

```python
set1 = {1, 2, 3, 4, 5, 6}
set2 = {4, 5, 6, 7, 8, 9}
difference_set = set1.difference(set2)
print(difference_set)
```

You can also update a set with the difference using difference_update:

```python
set1.difference_update(set2)
print(set1)
```

To get elements unique to the second set, use:

```python
set2.difference(set1)
print(set2.difference(set1))
```

### Symmetric Difference
Symmetric difference returns elements that are unique to each set, excluding common elements.

```python
set1 = {1, 2, 3, 4, 5, 6}
set2 = {4, 5, 6, 7, 8, 9}
symmetric_diff = set1.symmetric_difference(set2)
print(symmetric_diff)
```

#### Set Methods: Subset and Superset
To check if a set is a subset or superset of another set, use the following methods:

```python
# Define two sets
set1 = {1, 2, 3}      # Set1 ke elements
set2 = {3, 4, 5}      # Set2 ke elements

# Check if set1 is a subset of set2
# issubset() returns True if all elements of set1 are in set2
print(set1.issubset(set2))   # False, kyunki 1 aur 2 set2 me nahi hain

# Check if set1 is a superset of set2
# issuperset() returns True if set1 contains all elements of set2
print(set1.issuperset(set2)) # False, kyunki 4 aur 5 set1 me nahi hain
```

If all elements of set1 are in set2, issubset returns True. If set1 contains all elements of set2, issuperset returns True.

#### Removing Duplicates from a List Using Sets
To quickly remove duplicates from a list, convert it to a set:

```python
my_list = [1, 2, 2, 3, 4, 4, 5]
unique_elements = set(my_list)
print(unique_elements)
```

#### Practical Example: Counting Unique Words in Text
Suppose you have a sentence and want to count the number of unique words.

```python
text = "In this tutorial we are discussing about sets."
words = text.split()
unique_words = set(words)
print(unique_words)
print(len(unique_words))
```
