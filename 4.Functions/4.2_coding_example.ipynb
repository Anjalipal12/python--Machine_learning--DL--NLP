#### Temperature Conversion Function
The first example demonstrates how to create a function for temperature conversion between Celsius and Fahrenheit.

```python
def convert_temperature(temperature, unit):
    """
    Converts temperature between Celsius and Fahrenheit.
    """
    if unit == 'C':
        return (temperature * 9/5) + 32
    elif unit == 'F':
        return (temperature - 32) * 5/9
    else:
        return None
```

```python
print(convert_temperature(25, 'C'))  # Output: 77.0
print(convert_temperature(77, 'F'))  # Output: 25.0
```

#### Password Strength Checker
The next example is a password strength checker function. It checks if a password is strong based on several criteria: minimum length, presence of digits, lowercase and uppercase letters, and special characters.

```python
def is_strong_password(password):
    """
    Checks if the password is strong or not.
    """
    if len(password) < 8:
        return False
    if not any(char.isdigit() for char in password):
        return False
    if not any(char.islower() for char in password):
        return False
    if not any(char.isupper() for char in password):
        return False
    special_characters = '!@#$%^&*()-_+=[]{}|;:,.<>?/\\'
    if not any(char in special_characters for char in password):
        return False
    return True
```

```python
print(is_strong_password('AbAbaBaB'))  # Output: False
print(is_strong_password('Str0ngPwd!'))  # Output: True
```

#### Shopping Cart Total Cost Calculation
This example shows how to calculate the total cost of items in a shopping cart, where the cart is represented as a list of dictionaries.

```python
def calculate_total_cost(cart):
    total_cost = 0
    for item in cart:
        total_cost += item['price'] * item['quantity']
    return total_cost
```

#### Example Cart Data
Here is an example of how the cart data might look and how to use the function:

```python
cart = [
    {'name': 'Apple', 'price': 0.5, 'quantity': 4},
    {'name': 'Banana', 'price': 0.3, 'quantity': 6},
    {'name': 'Orange', 'price': 0.7, 'quantity': 3}
]
total_cost = calculate_total_cost(cart)
print(total_cost)  # Output: 5.889
```

#### Palindrome Checker
A function to check if a string is a palindrome, ignoring spaces and case.

```python
def is_palindrome(s):
    s = s.lower().replace(' ', '')
    return s == s[::-1]
```

Example Usage
Testing the palindrome checker with different strings:

```python
print(is_palindrome('A man a plan a canal Panama'))  # Output: True
print(is_palindrome('hello'))  # Output: False
```

#### Factorial Calculation Using Recursion
This example demonstrates how to calculate the factorial of a number using recursion.

```python
def factorial(n):
    if n == 0:
        return 1
    else:
        return n * factorial(n - 1)
```

Example Usage
Calculating the factorial of 5 and 6:

```python
print(factorial(5))  # Output: 120
print(factorial(6))  # Output: 720
```

#### Word Frequency Counter from a File
This function reads a file and counts the frequency of each word, demonstrating file operations and dictionary usage.

```python
def count_word_frequency(file_path):
    word_count = {}
    with open(file_path, 'r') as file:
        for line in file:
            words = line.split()
            for word in words:
                word = word.lower().strip('.,:;!?/\\')
                word_count[word] = word_count.get(word, 0) + 1
    return word_count
```

Step-by-step Samjho:

word_count = {}
→ Ek empty dictionary banayi — yahan har word aur uska count store hoga.
Example: {'hello': 2, 'world': 1}

with open(file_path, 'r') as file:
→ File open karte hain read mode mein ('r' = read).
→ Example: agar file me likha hai "Hello world Hello"

for line in file:
→ File ke har line ko ek ek karke padhta hai.

words = line.split()
→ Line ko alag-alag words mein todta hai.
Example: "Hello world" → ['Hello', 'world']

for word in words:
→ Ab har word ke liye loop chalega.

word = word.lower().strip('.,:;!?/\\')
→ Word ko lowercase mein badalta hai (so Hello → hello)
→ Aur punctuation (.,!? etc) hata deta hai.

word_count[word] = word_count.get(word, 0) + 1
→ Agar word pahle se dictionary me hai → count +1 karo
→ Agar nahi hai → 0 se start karke +1 kar do.

#### Example Usage
Assuming 'sample.txt' is present in the working directory:

```python
file_path = 'sample.txt'
word_frequency = count_word_frequency(file_path)
print(word_frequency)
```

#### Email Address Validation
The final example demonstrates how to validate an email address using regular expressions.

```python
import re

def is_valid_email(email):
    pattern = r'^[A-Za-z0-9+_.-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}$'
    return re.match(pattern, email) is not None
```

#### Example Usage
Testing the email validation function:

```python
print(is_valid_email('username@gmail.com'))  # Output: True
print(is_valid_email('invalid-email'))  # Output: False
```
