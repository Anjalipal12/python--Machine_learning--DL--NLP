### Function Copy, Closures, and Decorators in Python
#### Introduction to Advanced Python Concepts
In this lecture, we continue our discussion on Python, focusing on advanced concepts such as function copy, closures, and decorators. These topics are essential for understanding how to modify and extend the behavior of functions and methods in Python.

#### Understanding Function Copy
Function copy in Python refers to assigning a function to another variable. This allows the function to be called using the new variable, even if the original function is deleted.

```python
def welcome():
    return "Welcome to the Advanced Python course"
```

```python
well = welcome
print(well())
del welcome
print(well())
```

Even after deleting the original welcome function, the copied function well continues to work, demonstrating that the function object is preserved in the new variable.

#### Closures in Python
A closure is a function defined inside another function, which can access variables from the enclosing function's scope. This is similar to concepts in other languages like C# and Java.

```python
def main_welcome():
    message = "Welcome"
    def sub_welcome():
        print("Welcome to the Advanced Python course")
        print("Please learn this concepts properly")
        print(message)
    return sub_welcome
```

```python
closure_func = main_welcome()
closure_func()
```

In the above example, sub_welcome is a closure that accesses the message variable from its enclosing function. Closures can also accept functions as arguments and use them within the inner function.

```python
def main_welcome(func):
    def sub_welcome():
        print("Welcome to the Advanced Python course")
        print("Please learn this concepts properly")
        func("Welcome everyone to this tutorial.")
    return sub_welcome
```

```python
closure_func = main_welcome(print)
closure_func()
```

Closures can also be used to pass in-built functions like print or len as arguments, allowing for dynamic behavior within the inner function.

```python
def main_welcome(func, lst):
    def sub_welcome():
        print("Welcome to the Advanced Python course")
        print(func(lst))
    return sub_welcome
```

Tip: return sub_welcome → function ko return karna,
return sub_welcome() → function ko turant run karna.

```python
closure_func = main_welcome(len, [1,2,3,4,5])
closure_func()
```

#### Introduction to Decorators
Decorators are a powerful and flexible feature in Python that allow you to modify the behavior of a function or class method. They are commonly used to add functionality to functions or methods without modifying the actual code.

```python
def main_welcome(func):
    def sub_welcome():
        print("Welcome to the Advanced Python course")
        func()
        print("Please learn this concepts properly")
    return sub_welcome
```

```python
def course_introduction():
    print("This is an advanced Python course")
```

```python
@main_welcome
def course_introduction():
    print("This is an advanced Python course")
```

```python
course_introduction()
```

By using the @main_welcome decorator, the course_introduction function is passed as a parameter to the decorator, and the additional functionality is executed before and after the original function.

#### Another Example: Decorator with Pre- and Post-Functionality
Decorators can be used to execute code before and after the decorated function is called.

```python
def my_decorator(func):
    def wrapper():
        print("Something is happening before the function is called.")
        func()
        print("Something is happening after the function is called.")
    return wrapper
```

```python
@my_decorator
def say_hello():
    print("Hello")
```

```python
say_hello()
```

When say_hello() is called, the decorator ensures that messages are printed before and after the main function's output.

#### Decorators with Arguments
Decorators can also accept arguments, allowing for more flexible and dynamic behavior.

```python
def greet_decorator(msg):  # decorator ko argument mila
    def decorator(func):
        def wrapper():
            print(msg)   # argument ka use
            func()
        return wrapper
    return decorator

@greet_decorator("Hello Students!")  # decorator ke saath argument
def course():
    print("Welcome to Python course")

course()
```

Extra behavior add karna

Kabhi hume ek function me extra kaam karna hota hai, jaise logging, message print karna, validation check, time measure karna, etc.

Decorator allow karta hai ye kaam original function ko modify kiye bina karne ka.

2️⃣ Code ko clean aur reusable banana

Agar har function me alag-alag message ya check likhna shuru karenge, code messy ho jayega.

Decorator use karke ek hi jagah logic define karte hain, aur multiple functions me reuse kar sakte hain.

3️⃣ Arguments ke saath flexibility

Agar decorator arguments accept kare (jaise "Hello Students!"), to har function ke liye different behavior define karna easy ho jata hai.

```python
def repeat(n):
    def decorator(func):
        def wrapper(*args, **kwargs):
            for _ in range(n):
                func(*args, **kwargs)
        return wrapper
    return decorator
@repeat(3)
def say_hello():
    print("Hello")

say_hello()
```

In this example, the say_hello function is executed three times due to the decorator's argument.

Decorators are a powerful tool in Python for extending and modifying the behavior of functions and methods. They allow for clean, reusable, and flexible code, and are widely used in frameworks such as Flask and Django.
