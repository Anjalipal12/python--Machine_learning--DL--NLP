```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.model_selection import StratifiedShuffleSplit

from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import OrdinalEncoder
from sklearn.preprocessing import OneHotEncoder

from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer

from sklearn.linear_model import LinearRegression

from sklearn.metrics import mean_squared_error, r2_score

house = pd.read_csv('housing.csv')

X = house.drop('median_house_value', axis = 1)
y = house.median_house_value

X_train,X_test, y_train, y_test = train_test_split(X,y,test_size= 0.1, random_state = 121)

num_pipeleine = Pipeline([
    ('impute', SimpleImputer(strategy = 'median')),
    ('scaler', StandardScaler())

])

cat_pipeline = Pipeline([
    ('impute', SimpleImputer(strategy = "most_frequent")),
    ('encoding', OrdinalEncoder())
])

num_col = X.select_dtypes('float').columns
cat_col = X.select_dtypes('object').columns

final_pipeline = ColumnTransformer([
    ('num_pipeline', num_pipeleine, num_col),
    ('cat_pipeline', cat_pipeline, cat_col)
])

final_pipeline.fit(X_train)
X_train_tr = final_pipeline.transform(X_train)
X_test_tr = final_pipeline.transform(X_test)

model = LinearRegression()
model.fit(X_train_tr, y_train)

y_train_hat = model.predict(X_train_tr)
y_test_hat = model.predict(X_test_tr)

print(f"Training_error: {mean_squared_error(y_train, y_train_hat):.2f}")
print(f"Training_accuracy: {r2_score(y_train, y_train_hat):.2f}")
print()
print(f"Testing_error: {mean_squared_error(y_test, y_test_hat):.2f}")
print(f"Testiing_accuracy: {r2_score(y_test,y_test_hat):.2f}")
```
