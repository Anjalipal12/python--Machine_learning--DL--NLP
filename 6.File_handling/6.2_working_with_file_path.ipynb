### Working With File Paths
Creating a New Directory
The next task is to create a new directory. To do this, first import the os module. Suppose the new directory is named package. The goal is to create a package folder in the same working location, specifically inside the file handling folder.

```python
import os
new_directory = "package"
os.mkdir(new_directory)
print(f"Directory {new_directory} is created.")
```

After executing this code, the directory package is created. You will be able to see the package folder in the specified location. This demonstrates how to create a new directory using the os module.

#### Listing Files and Directories
To list all files and folders in the current working directory, use the following approach:

```python
items = os.listdir('.')
print(items)
```

#### Joining Paths
A very important concept is joining paths. This is a key feature of the os module, as file paths differ between operating systems. Using os.path.join ensures compatibility.

```python
dir_name = "folder"
file_name = "file.txt"
full_path = os.path.join(os.getcwd() ,dir_name , file_name)
print(full_path)
```
