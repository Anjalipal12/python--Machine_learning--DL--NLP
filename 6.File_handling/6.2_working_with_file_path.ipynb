### Working With File Paths
Creating a New Directory
The next task is to create a new directory. To do this, first import the os module. Suppose the new directory is named package. The goal is to create a package folder in the same working location, specifically inside the file handling folder.

```python
import os
new_directory = "package"
os.mkdir(new_directory)
print(f"Directory {new_directory} is created.")
```

After executing this code, the directory package is created. You will be able to see the package folder in the specified location. This demonstrates how to create a new directory using the os module.

#### Listing Files and Directories
To list all files and folders in the current working directory, use the following approach:

```python
items = os.listdir('.')
print(items)
```

#### Joining Paths
A very important concept is joining paths. This is a key feature of the os module, as file paths differ between operating systems. Using os.path.join ensures compatibility.

```python
dir_name = "folder"
file_name = "file.txt"
full_path = os.path.join(os.getcwd() ,dir_name , file_name)
print(full_path)
```

This code creates a path like folder/file.txt, representing a file inside a folder. To get the full path including the current working directory, use the following:

```python
import os
full_path = os.path.join(os.getcwd(), directory_name, file_name)
print(full_path)
```

This prints the complete path. If only the relative path is needed, use just the directory and file name variables.
#### Checking If a Path Exists
When working with files, it is important to check if a file or directory exists before performing operations.

```python
example_file = "example1.txt"
if os.path.exists(example_file):
    print(f"The path {example_file} exists.")
else:
    print(f"The path {example_file} does not exist.")
```

If the file does not exist, you can create it using the open command or any other file creation method. This kind of condition is important when making assumptions about specific paths in your code.

#### Checking If a Path Is a File or Directory
There is another way to check whether a path is a file or a directory.

```python
path = "example.txt"
if os.path.isfile(path):
    print(f"The path {path} is a file.")
elif os.path.isdir(path):
    print(f"The path {path} is a directory.")
else:
    print(f"The path {path} is neither a file nor a directory.")
```

This code checks if the path is a file, a directory, or neither. For example, if example.txt is a file, it will print that the path is a file.

#### Relative and Absolute Paths
A relative path refers to a file inside a specific folder, such as file.txt inside a folder. An absolute path is the complete path from the root directory.

```python
relative_path = "example.txt"
absolute_path = os.path.abspath(relative_path)
print(absolute_path)
```

This prints the full absolute path of the file. Alternatively, you can join components to build the path, but using os.path.abspath is often more convenient.
