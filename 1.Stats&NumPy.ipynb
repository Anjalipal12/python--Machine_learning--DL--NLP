#### Spreadness / Dispertion of Data

    Range
    Std
    Var

```python
# Dispertion --> data kitni baar faili hua ha
```

```python
import numpy as np
```

```python
ages = np.random.randint(25,35,100) # 100 random numbers from 25 to 35
ages
```

```python
mean = np.mean(ages)
median = np.median(ages)
print(mean,median,sep='\n')
```

Std

```python
import matplotlib . pyplot as plt
```

```python
plt.plot(ages , 'ro') # x = index , y = ages , red circles
plt.show()
```

```python
plt.plot(ages,'ro')
plt.plot([0,100],[mean,mean])
plt.show()
```

```python
ages
```

dev = mean - data ----> distance of each age from mean

```python
dev = [abs(mean-i) for i in ages]
print(dev)
# abs ---> convert negative value to positive
```

```python
np.mean(dev)   # avergae distance from mean to ages
```

```python
std = np.std(ages)  # overall spread
std
```

```python
plt.plot(ages,'ro')
plt.plot([0,100],[mean,mean],label = 'Mean')
plt.plot([0,100],[mean+std , mean+std],label = 'Mean+std')
plt.plot([0,100],[mean-std , mean-std],label ='mean-std')
plt.legend() # graph mein dikha deta ha
plt.show()
```

variance = spread of ages squared

```python
var = np.var(ages)
var
```

```python
plt.plot(ages,'ro')
plt.plot([0,100],[mean,mean], label='mean')
plt.plot([0,100],[mean+var , mean+var], label='mean + var')
plt.plot([0,100],[mean-var,mean-var],label='mean - var')
plt.legend()
plt.show()
```

```python
plt.plot(ages,'ro')
plt.plot([0,100],[mean,mean], label = 'mean')
plt.plot([0,100],[mean+2*std,mean+2*std], label = 'mean+2*std')
plt.plot([0,100],[mean-2*std, mean-2*std],label = 'mean-2*std')
plt.legend()
plt.show()
```

```python
normal = np.random.normal(150,20,1000) # normal --> generate random data from real world
normal
```

```python
np.mean(normal)  # ye bell curve ka center ha (average) dikhata ha approx 150
```

```python
np.std(normal)  # curve ka width (spread) dikhata ha approx 20
```

plots


```python
plt.hist(normal , bins = 30 , ec ='k')  # ec = 'k' ---> bars ke edeges black ha
plt.show()                              # bins = 30 ka matlab hota ha ki jo blocks ha voh 30 use kiya ha
```

```python
plt.boxplot(normal) # outlier ko detect karna ka liya use hota ha boxplot
plt.show()
```

```python
quantiles = np.quantile(normal , [0.25 , 0.50 , 0.75]) # q1 = 0.25 , 0.50 = q2 , 0.75 = q3
quantiles
```

```python
q1 = quantiles[0]
q2 = quantiles[1]
q3 = quantiles[2]
```

```python
q1
```

```python
q2
```

```python
q3
```

```python
IQR = q3 - q1
```

```python
IQR  # middle 50 % ka spread ha jo 25.01 units ha 
     # focus on middle values
```

```python
normal  # full realistic data ka shape
```

```python
lower_bound = q1 - 1.5 * IQR
upper_bound = q3 - 1.5 * IQR
```

```python
lower_bound
```

```python
upper_bound
```

```python
outliers = normal[(normal < lower_bound) | (normal > upper_bound)]  # # outlier bahut choti yaah badi values , jo middle 50 % ke bahar ha
outliers
```
