1️⃣ Question: What is a normal distribution?

Answer:
A normal distribution is a continuous probability distribution that is symmetric and bell-shaped. Most data points lie around the mean, and probabilities decrease as we move away from the mean.

2️⃣ Question: How do you generate random data in Python following a normal distribution?

Answer:

import numpy as np
normal = np.random.normal(mean, std_dev, size)


Example:

normal = np.random.normal(150, 20, 1000)


mean = 150, std = 20, 1000 data points.

3️⃣ Question: How do you find the center and spread of the normal data?

Answer:

np.mean(normal)  # center
np.std(normal)   # spread


Mean ≈ center of bell curve.

Std ≈ width/spread of curve.

4️⃣ Question: How do you visualize a normal distribution?

Answer:

import matplotlib.pyplot as plt
plt.hist(normal, bins=30, ec='k')  # histogram
plt.show()
plt.boxplot(normal)                # boxplot
plt.show()

5️⃣ Question: How do you detect outliers using IQR?

Answer:

q1, q2, q3 = np.quantile(normal, [0.25, 0.50, 0.75])
IQR = q3 - q1
lower_bound = q1 - 1.5*IQR
upper_bound = q3 + 1.5*IQR
outliers = normal[(normal < lower_bound) | (normal > upper_bound)]


Values below lower_bound or above upper_bound are outliers.

6️⃣ Question: If mean = 150, std = 20, what is the probability of selecting a value >170?

Answer:

Step 1: Calculate Z-score:

z = (170 - 150)/20  # z = 1


Step 2: Use z-table or scipy.stats.norm:

from scipy.stats import norm
prob = 1 - norm.cdf(z)  # ≈ 0.1587


Probability ≈ 15.87%

### Skewness

1.left skewness         -----> mean < median (data - left skewed && negatively skewed) (chotti values jada affect ho rahi ha aur avg ko neecha khich rahi ha)

2.Right skewness   ------> mean > median

3. Zero Skewness ---> mean = median

1. Left-Skewed (Mean < Median):

Curve ka tail left side mein hai (chhoti values ki wajah se).

Mean chhoti values se neeche kheench jaata hai, isliye Mean < Median.

2. Right-Skewed (Mean > Median):

Curve ka tail right side mein hai (badi values ki wajah se).

Mean badi values se upar kheench jaata hai, isliye Mean > Median.

Rule yaad rakhne ke liye:

Tail ki taraf dekho → Mean usi side khicha hua hota hai, Median zyada stable hai.
