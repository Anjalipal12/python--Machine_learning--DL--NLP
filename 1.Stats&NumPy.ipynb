```python
ages = list(np.random.randint(19, 25, 100))
print(ages)
```

```python
sum(ages) / len(ages)
```

```python
mean = np.mean(ages)
mean
```

```python
median = np.median(ages)
median
```

```python
print("Without Outlier")
print(f"Mean: {mean}")
print(f"Median: {median}")
```

```python
country = np.random.choice(['aus', 'india', 'china', 'usa'], 100)
country
```

```python
data = {}

for i in country:
    if i in data:
        data[i] += 1
    else:
        data[i] = 1
print(data)
```

```python
data.items()
```

```python
m = max(list(data.values()))
m
```

```python
for i,j in data.items():
    print(i)
```

```python
for i,j in data.items():
    if j == m:
        print(i)
```
