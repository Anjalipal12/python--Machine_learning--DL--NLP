Logging is a crucial aspect of any application , providing a way to track events, errors and operational information. Python's built-in logging module offers a flexible framework for emitting log messages from Python programs. In this lesson, we will cover the basics of logging , including how to configure logging, log levels and best practices for using logging in Python applications

```python
import logging

## Configure the basics logging seetings
logging.basicConfig(level=logging.DEBUG) # giving indication that hi ia am developing a entire application

## log messages with severity levels .... if i had use different different levels so i will use this
logging.debug("This is debug message")
logging.info("This is an info messages")
logging.warning("This is a warning message")
logging.error("this is an error message")
logging.critical("This is a critical message")
```

## Logging Levels
Python's logging module has several log levels indicating the severity of events. The default levels are:
1. DEBUG: Detailed information , typically of interest only when diagnosing problems
2. INFO: Confirmation that things are working as expected.
3. WARNING: An indication that something unexpected happened or indicative of some problem in the near future (eg., 'disk space low'). The software is still working as expected.
4. ERROR: Due to a more serious problem, the software has not been able to perform some function.
5. CRITICAL: A very serious error, indicating that the program itself may be unable to continue running

```python
## focus more on Configuring logging ---> how the logging 
logging.basicConfig(
    filename='app.log',
    filemode='w',
    level=logging.DEBUG,
    format='%(asctime)s-%(name)s-%(levelname)s-%(message)s',
    datefmt='%Y-%m-%d %H:%M:%S'
)

# log messages with severity levels .... if i had use different different levels so i will use this
logging.debug("This is debug message")
logging.info("This is an info messages")
logging.warning("This is a warning message")
logging.error("this is an error message")
logging.critical("This is a critical message")
```
